LINE PLOT (selected only the first 100 rows)

```python
plt.figure(figsize=(10, 5))

plt.plot(df['customer_age'].head(100))

plt.title('Line Plot of Customer Age')
plt.xlabel('Index')
plt.ylabel('Customer Age')

plt.show()

```

```python
plt.figure(figsize=(10, 5))

plt.plot(df['customer_age'].head(100), marker='o')

plt.title('Line Plot of Customer Age')
plt.xlabel('Index')
plt.ylabel('Customer Age')

plt.show()
```

BAR CHART

```python
age_count = df['customer_age'].value_counts().sort_index()

plt.figure(figsize=(10, 5))

plt.bar(age_count.index, age_count.values)

plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Number of Records')

plt.show()
```

```python

plt.figure(figsize=(10, 8))

plt.barh(age_count.index, age_count.values,
         color='skyblue')

plt.title('Customer Age Distribution')
plt.xlabel('Number of Records')
plt.ylabel('Customer Age')

plt.show()
```

```python
age_count = df['customer_age'].value_counts().sort_index()

plt.figure(figsize=(10, 5))

plt.bar(age_count.index, age_count.values,
        color='lightgreen',
        edgecolor='black')

plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Number of Records')

plt.show()
```

SCATTER PLOT

```python
plt.figure(figsize=(10, 5))

plt.scatter(df['customer_age'].head(100),
            df['net_sales'].head(100))

plt.title('Customer Age vs Net Sales')
plt.xlabel('Customer Age')
plt.ylabel('Net Sales')

plt.show()
```

PIE CHART

```python
payment_count = df['payment_method'].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(payment_count.values,
        labels=payment_count.index,
        autopct='%1.1f%%'
        )

plt.title('Distribution of Payment Methods')

plt.show()

```

HISTOGRAM

```python
plt.figure(figsize=(8,5))

plt.hist(df['customer_age'], bins=20)

plt.title('Distribution of Customer Ages')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['customer_age'], bins=20, edgecolor='black')

plt.title('Distribution of Customer Ages')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')

plt.show()
```

DATA VISUALIZATION USING SEABORN

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

LINE PLOT

```python
plt.figure(figsize=(10, 5))

sns.lineplot(data=df.head(100), x=df.head(100).index,
             y='customer_age')

plt.title('Line Plot of Customer Age')
plt.xlabel('Index')
plt.ylabel('Customer Age')

plt.show()
```

BAR PLOT

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='customer_age')

plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Number of Records')

plt.show()
```

SCATTER PLOT

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(data=df.head(100),
                x='customer_age',
                y='net_sales')

plt.title('Customer Age vs Net Sales')
plt.xlabel('Customer Age')
plt.ylabel('Net Sales')

plt.show()
```

HISTOGRAM

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='customer_age', bins=20)

plt.title('Distribution of Customer Ages')
plt.xlabel('Customer Age')
plt.ylabel('Frequency')

plt.show()
```

BOX PLOT

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y='net_sales')

plt.title('Distribution of Net Sales')
plt.ylabel('Net Sales')

plt.show()
```
