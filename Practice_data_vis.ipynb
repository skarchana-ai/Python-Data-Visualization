# Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset("tips")
df
```

```python
#scatter plot
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex",
    style="day"
)

plt.show()
```

```python
#barplot
sns.barplot(
    data=df,
    x="total_bill",y="tip"
)

plt.show()
```

```python
#count plot
sns.countplot(
    data=df,
    x="day"
)

plt.show()
```

```python
#histogram
sns.histplot(
    data=df,
    x="day"

)
plt.show()
```

```python
#kde-kernel distribution estimation
sns.kdeplot(
    data=df,
    x="size"
)
plt.show()
```

```python
#histogram + kde
sns.histplot(
    data=df,
    x="day",
    kde=True

)
plt.show()
```

```python
#violin plot
sns.violinplot(
    data=df,
    x="tip"
)

plt.show()
```

```python
#box plot

sns.boxplot(
data=df,
x="tip"
)
plt.show()
```

```python
#line plot
sns.lineplot(
    data=df,
    x="tip",
    y="total_bill"
)

plt.show()
```

```python
x=df['total_bill']
y=df['tip']

plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.plot(x,y,marker='s')
plt.show()
```

```python
#Scatter plot

x=df['total_bill']
y=df['tip']

plt.scatter(x,y)
```

```python
#pie chart
plt.pie(x,labels=df['total_bill'],autopct='%1.1f%%')
```

```python
#histogram
plt.hist(x,bins =5,edgecolor='black')
```

```python
#sub plots

x=df['total_bill']
y=df['tip']

#1st graph
plt.subplot(2,2,1)
plt.plot(x,y)

#2nd graph
plt.subplot(2,2,2)
plt.bar(x,y)

#3rd graph
plt.subplot(2,2,3)
plt.scatter(x,y)

#4th graph
plt.subplot(2,2,4)
plt.hist(x,bins=5,edgecolor='black')


plt.show()
```
